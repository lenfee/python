```python
import re

# 1. 匹配数字、空白字符和完整单词 cat
text = "cat on mat; later: cats and dogs"
print(r"\d+:", re.findall(r"\d+", text))
print(r"\s+:", re.findall(r"\s+", text))
print(r"\bcat\b:", re.findall(r"\bcat\b", text))

# 2. | 表示选择分支；匹配区分大小写
print("cat|dog:", re.findall(r"cat|dog", "cat dog bird CAT"))
first_short = re.findall(r"a|aa|aaa", "aaaa")
first_long = re.findall(r"aaa|aa|a", "aaaa")
print("a|aa|aaa:", first_short)
print("aaa|aa|a:", first_long)
# 引擎按分支顺序尝试：先写 a 会先匹配单个字符；先写 aaa 则优先匹配最长分支。

# 3. * 为 0 次或多次，+ 为 1 次或多次，? 为 0 次或 1 次；
#    {2} 为恰好 2 次，{1,2} 为 1 到 2 次。
text = "ct cat caat caaat"
for pattern in (r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"):
    print(f"{pattern}: {re.findall(pattern, text)}")
```

```python
import re

# 1. 捕获邮箱用户名和域名
email_text = "联系：alice@example.com；备选：bob@test.org"
email_pattern = r"([\w.+-]+)@([\w.-]+\.[A-Za-z]{2,})"
print("邮箱用户名和域名：", re.findall(email_pattern, email_text))

# 2. 使用反向引用 \1 匹配连续重复的单词
words_text = "cat cat, dog cat, bird bird"
print("连续重复的单词：", re.findall(r"\b(\w+)\s+\1\b", words_text))

# 3. 用捕获组将 YYYY-MM-DD 转成 MM-DD-YYYY，并检查行数不变
date_text = "2024-05-01\n2024/05/02\n2024.05.03"
date_pattern = r"\b(\d{4})[-/.](\d{2})[-/.](\d{2})\b"
dates_reformatted = re.sub(date_pattern, r"\2-\3-\1", date_text)
print("转换前日期：\n" + date_text)
print("转换后日期：\n" + dates_reformatted)
print("行数：", len(date_text.splitlines()), "->", len(dates_reformatted.splitlines()))

# 4. 将连字符或点分隔的电话号码统一成 (区号) 号码-号码
phone_text = "联系我：123-456-7890 或 987.654.3210"
phone_pattern = r"(?<!\d)(\d{3})[-.](\d{3})[-.](\d{4})(?!\d)"
phones_reformatted = re.sub(phone_pattern, r"(\1) \2-\3", phone_text)
print("转换前电话：", phone_text)
print("转换后电话：", phones_reformatted)
```

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

# 1. 构建解析树并输出页面标题
soup = BeautifulSoup(html_doc, "html.parser")
print("页面标题：", soup.title.get_text())
print("h1 标题：", soup.h1.get_text())

# 2. 输出所有 class="course" 的段落及所有链接的文字和 href 属性
print("课程段落：")
for p in soup.find_all("p", class_="course"):
    print(p.get_text())

print("链接：")
for a in soup.find_all("a"):
    print(a.get_text(), "-", a["href"])

# 3. 使用 get_text() 提取文档、段落和链接文本
print("整个文档可见文本：", soup.get_text())
print("单个段落文本：", soup.find("p", class_="course").get_text())
print("单个链接文本：", soup.find("a").get_text())
```
