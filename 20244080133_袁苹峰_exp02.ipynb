```python
import numpy as np

# 1. 不同类型的 NumPy 数组
int_array = np.array([1, 2, 3])
float_array = np.array([1.0, 2.5, 3.0])
bool_array = np.array([True, False, True])

for name, array in (("整数数组", int_array), ("浮点数数组", float_array), ("布尔数组", bool_array)):
    print(f"{name}: dtype={array.dtype}, shape={array.shape}, ndim={array.ndim}")

print("NumPy 数组通常只保存同一种数据类型，便于高效进行数值运算；Python 列表可以混合不同类型，长度也可变。")

# 2. 创建并改变数组形状
values = np.arange(12)
array_3x4 = values.reshape(3, 4)
array_2x3x2 = values.reshape(2, 3, 2)
print("原数组:", values)
print("形状 (3, 4):\n", array_3x4)
print("形状 (2, 3, 2):\n", array_2x3x2)
print("reshape 要求新形状各维度乘积等于元素总数。")

# 3. 索引和布尔筛选
second_row = array_3x4[1, :]
columns_1_to_3 = array_3x4[:, 0:3]
greater_than_6 = array_3x4[array_3x4 > 6]
for name, result in (("第 2 行", second_row), ("第 1 至第 3 列", columns_1_to_3), ("大于 6 的元素", greater_than_6)):
    print(f"{name}: {result}, shape={result.shape}")

# 4. 组合条件筛选
numbers = np.arange(10)
less_than_2_or_greater_than_7 = numbers[np.logical_or(numbers < 2, numbers > 7)]
greater_than_3_and_less_than_7 = numbers[np.logical_and(numbers > 3, numbers < 7)]
print("小于 2 或大于 7:", less_than_2_or_greater_than_7)
print("大于 3 且小于 7:", greater_than_3_and_less_than_7)

# 5. 逐元素乘法与矩阵乘法
matrix_a = np.array([[1, 2], [3, 4]])
matrix_b = np.array([[5, 6], [7, 8]])
print("A * B（逐元素乘法）:\n", matrix_a * matrix_b)
print("np.multiply(A, B)（逐元素乘法）:\n", np.multiply(matrix_a, matrix_b))
print("A @ B（矩阵乘法）:\n", matrix_a @ matrix_b)
print("np.matmul(A, B)（矩阵乘法）:\n", np.matmul(matrix_a, matrix_b))
```

```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

# 1. 生成并限制模拟成绩，再计算统计量
scores = np.clip(rng.normal(loc=70, scale=10, size=1000), 0, 100)
print("样本数:", scores.size)
print("最小值:", scores.min())
print("最大值:", scores.max())
print("均值:", scores.mean())
print("标准差:", scores.std())
print("极差:", np.ptp(scores))

# 2. 从 1 到 10 不放回抽取 5 个数
population = np.arange(1, 11)
uniform_sample = rng.choice(population, size=5, replace=False)
probabilities = population / population.sum()
weighted_sample = rng.choice(population, size=5, replace=False, p=probabilities)
print("均匀抽样:", uniform_sample)
print("按数值大小成比例加权抽样:", weighted_sample)

# 3. permutation 返回打乱后的新数组；shuffle 原地打乱数组
original_for_permutation = np.arange(1, 11)
permuted = rng.permutation(original_for_permutation)
print("permutation 操作前:", original_for_permutation)
print("permutation 返回的新数组:", permuted)
print("permutation 操作后原数组:", original_for_permutation)

shuffled_in_place = np.arange(1, 11)
print("shuffle 操作前:", shuffled_in_place)
rng.shuffle(shuffled_in_place)
print("shuffle 操作后原数组:", shuffled_in_place)

# 4. 设置缺失值并比较普通均值与忽略缺失值后的均值
scores_with_missing = scores.copy()
scores_with_missing[[10, 20, 30]] = np.nan
print("缺失值个数:", np.isnan(scores_with_missing).sum())
print("含缺失值时的普通均值:", np.mean(scores_with_missing))
print("忽略缺失值后的均值:", np.nanmean(scores_with_missing))
print("np.mean 遇到 NaN 会得到 NaN；np.nanmean 会跳过 NaN。")

# 5. 建立仅保存在内存中的表格，供实验 4 使用，不写入磁盘
grades_table = pd.DataFrame({
    "student_id": np.arange(1, scores_with_missing.size + 1),
    "score": scores_with_missing,
})
print("内存表格变量 grades_table（前 5 行）:")
print(grades_table.head())
print("表格行数:", len(grades_table))
```

```python
import pandas as pd

# 1. Build the DataFrame, use names as the index, and calculate total and average.
records = {
    "Name": ["Alice", "Bob", "Carol", "David", "Eva", "Frank"],
    "Group": ["A", "A", "B", "B", "A", "B"],
    "Math": [88, 72, 91, 65, 79, 84],
    "English": [90, 68, 85, 73, 92, 77],
    "Python": [95, 80, 89, 70, 86, 82],
}

scores = pd.DataFrame(records).set_index("Name")
subjects = ["Math", "English", "Python"]
scores["Total"] = scores[subjects].sum(axis=1)
scores["Average"] = scores[subjects].mean(axis=1)

print("Row index:", scores.index.tolist())
print("Columns:", scores.columns.tolist())
print("First 3 rows:")
print(scores.head(3))
print("Basic statistics:")
print(scores.describe())

# 2. Read Carol with loc and iloc; select students whose average is at least 80.
print("\nCarol with loc:")
print(scores.loc["Carol"])
print("\nCarol with iloc:")
print(scores.iloc[scores.index.get_loc("Carol")])
print("\nStudents with average >= 80:")
print(scores.loc[scores["Average"] >= 80])

# 3. Calculate subject means and each student's mean across the three subjects.
print("\nMean by subject:")
print(scores[subjects].mean(axis=0))
print("\nMean by student:")
print(scores[subjects].mean(axis=1))

# 4. Find each student's highest score, subject summaries, and Math ranking.
scores["Highest"] = scores[subjects].apply(max, axis=1)
print("\nHighest score by student:")
print(scores["Highest"])
print("\nSubject min, mean, and max:")
print(scores[subjects].agg(["min", "mean", "max"]))
scores["Math_Rank"] = scores["Math"].rank(ascending=False, method="min").astype(int)
print("\nMath ranking (1 is highest):")
print(scores["Math_Rank"].sort_values())

# 5. Summarize group size and means; transform each group's Python mean to each row.
score_by_group = scores.groupby("Group")[subjects].mean()
score_by_group.insert(0, "Count", scores.groupby("Group").size())
print("\nGroup size and subject means:")
print(score_by_group)
scores["Group_Python_Average"] = scores.groupby("Group")["Python"].transform("mean")
print("\nGroup Python mean attached to each student:")
print(scores[["Group", "Python", "Group_Python_Average"]])

# 6. Keep groups whose mean Python score is at least 80.
qualified_groups = scores.groupby("Group").filter(lambda group: group["Python"].mean() >= 80)
print("\nGroups with Python mean >= 80 and their students:")
print(qualified_groups[["Group", "Python", "Group_Python_Average"]])

# 7. Calculate covariance and Pearson correlation matrices for the three subjects.
print("\nCovariance matrix:")
print(scores[subjects].cov())
print("\nPearson correlation matrix:")
print(scores[subjects].corr())
print("\nMath and Python have positive covariance and a strong positive correlation in this sample. Covariance magnitude depends on score scales; correlation is scale-independent and ranges from -1 to 1.")
```
