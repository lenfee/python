```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m


# 测试三组数字
print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
```

```python
def euler_limit(n):
    return (1 + 1 / n) ** n

#test
print(euler_limit(1))
print(euler_limit(10))
print(euler_limit(1000))
print(euler_limit(100000))
```

```python
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1  # 初始 0! = 1
    for k in range(n):
        total += 1 / fact
        fact *= (k + 1)
    return total

#test
print(euler_infinite_sum(0))  # 0.0
print(euler_infinite_sum(1))  # 1.0
print(euler_infinite_sum(4))  # 1 + 1 + 1/2 + 1/6 ≈ 2.6666666666666665
```

```python
import math

def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    fact = 1   # 0!
    k = 0
    while abs(true_e - total) > epsilon:
        total += 1 / fact
        k += 1
        fact *= k
    return total

#test
print(euler_approx(0.1))
print(euler_approx(0.001))
print(euler_approx(1e-6))
```

```python
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        val = euler_infinite_sum(k)
        print(val)

#test
print_euler_sum_table(4)
```

测试整数属性

```python
def is_power_of_2(n):
    while n % 2 == 0:
        n = n // 2
    return n == 1

def is_power(b, n):
    if n == 1:
        return True
    while n % b == 0:
        n = n // b
    return n == 1

#test
#1.
print(is_power_of_2(1))   # True  (2^0)
print(is_power_of_2(8))   # True
print(is_power_of_2(6))   # False

#2.
print(is_power(3, 9))   # True 3^2
print(is_power(5, 125)) # True
print(is_power(2, 7))   # False
print(is_power(4,1))    # True
```
