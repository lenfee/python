```python
def is_palindrome(text: str) -> bool:
    cleaned = "".join(char.lower() for char in text if char.isalnum())
    return cleaned == cleaned[::-1]

print(is_palindrome("T A C O cat"))
print(is_palindrome("hello"))
```

```python
def is_abecedarian(text: str) -> bool:
    letters = text.replace(" ", "").lower()
    return letters == "".join(sorted(letters))

print(is_abecedarian("adder"))
print(is_abecedarian("dog"))
print(is_abecedarian("abcd efgh xyz"))
```

```python
def remove_vowels(text: str) -> str:
    vowels = "aeiou"
    return "".join(char for char in text if char.lower() not in vowels)

print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("aeiou"))
```

```python
def list_reverse(items):
    if not isinstance(items, list):
        raise TypeError("items must be a list")
    return items[::-1]


def is_sorted(seq):
    return all(seq[index] <= seq[index + 1] for index in range(len(seq) - 1))


print(list_reverse([1, 2, 3]))
print(is_sorted([1, 2, 2, 3]))
print(is_sorted([3, 1, 2]))
```

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("t must be a tuple")
    if not all(isinstance(value, (int, float)) and not isinstance(value, bool) for value in t):
        raise TypeError("all elements of t must be integers or floats")
    if not isinstance(s, (int, float)) or isinstance(s, bool):
        raise TypeError("s must be an integer or float")
    return tuple(value * s for value in t)


def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("x and y must be tuples")
    if not all(isinstance(value, (int, float)) and not isinstance(value, bool) for value in x + y):
        raise TypeError("vector elements must be integers or floats")
    if len(x) != len(y):
        raise ValueError("vectors must have the same length")
    return sum(a * b for a, b in zip(x, y))


def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    if mx and any(len(row) != len(mx[0]) for row in mx):
        return False
    return all(
        isinstance(value, (int, float)) and not isinstance(value, bool)
        for row in mx
        for value in row
    )


def mx_vec_mult(mx, vec):
    if not isinstance(mx, tuple):
        raise TypeError("mx must be a tuple of tuples")
    if not check_valid_mx(mx):
        raise ValueError("mx must be a valid rectangular numeric matrix")
    if not isinstance(vec, tuple):
        raise TypeError("vec must be a tuple")
    if not all(isinstance(value, (int, float)) and not isinstance(value, bool) for value in vec):
        raise TypeError("vector elements must be integers or floats")
    if mx and len(mx[0]) != len(vec):
        raise ValueError("matrix columns must match vector length")
    return tuple(sum(value * element for value, element in zip(row, vec)) for row in mx)


print(vec_scalar_mult((1, 2, 3), 2))
print(vec_inner_product((1, 2, 3), (4, 5, 6)))
my_mx = ((1, 2), (3, 4))
print(check_valid_mx(my_mx))
print(mx_vec_mult(my_mx, (5, 6)))
```

```python
class Vector:
    def __init__(self, dimension, elements=None):
        if not isinstance(dimension, int) or isinstance(dimension, bool):
            raise TypeError("dimension must be an integer")
        if dimension < 0:
            raise ValueError("dimension cannot be negative")
        if elements is None:
            elements = (0,) * dimension
        if not isinstance(elements, (list, tuple)):
            raise TypeError("elements must be a list or tuple")
        if len(elements) != dimension:
            raise ValueError("dimension must match the number of elements")
        if not all(isinstance(value, (int, float)) and not isinstance(value, bool) for value in elements):
            raise TypeError("vector elements must be integers or floats")

        self.dimension = dimension
        self.elements = tuple(elements)

    def _check_other(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dimension != other.dimension:
            raise ValueError("vectors must have the same dimension")
        return other

    def __eq__(self, other):
        other = self._check_other(other)
        if other is NotImplemented:
            return NotImplemented
        return self.elements == other.elements

    def __lt__(self, other):
        other = self._check_other(other)
        if other is NotImplemented:
            return NotImplemented
        return self.elements < other.elements

    def __le__(self, other):
        other = self._check_other(other)
        if other is NotImplemented:
            return NotImplemented
        return self.elements <= other.elements

    def __gt__(self, other):
        other = self._check_other(other)
        if other is NotImplemented:
            return NotImplemented
        return self.elements > other.elements

    def __ge__(self, other):
        other = self._check_other(other)
        if other is NotImplemented:
            return NotImplemented
        return self.elements >= other.elements

    def dot(self, other):
        other = self._check_other(other)
        if other is NotImplemented:
            raise TypeError("dot expects another Vector")
        return sum(a * b for a, b in zip(self.elements, other.elements))

    def __mul__(self, other):
        if isinstance(other, Vector):
            other = self._check_other(other)
            return Vector(self.dimension, tuple(a * b for a, b in zip(self.elements, other.elements)))
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(self.dimension, tuple(value * other for value in self.elements))
        return NotImplemented

    def __rmul__(self, other):
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return self * other
        return NotImplemented

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"


v = Vector(2, (2, 3))
w = Vector(2, (4, 5))
print(v < w)
print(v.dot(w))
print(2 * v)
print(v * 2)
print(v * w)
```
