```python
# 1
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

# 2
s_for = 0
for i in range(1, 101):
    s_for += i
print("for求和：", s_for)

s_while = 0
i = 1
while i <= 100:
    s_while += i
    i += 1
print("while求和：", s_while)

# 3 迭代阶乘
def fact_iter(n):
    res = 1
    for i in range(2, n+1):
        res *= i
    return res

# 递归阶乘
def fact_rec(n):
    if n == 0 or n == 1:
        return 1
    return n * fact_rec(n - 1)

for x in [0,1,5,10]:
    print(x, fact_iter(x), fact_rec(x))

# 4
def summary(values):
    minv = min(values)
    maxv = max(values)
    avg = sum(values) / len(values)
    return minv, maxv, avg

lst = [1,3,5,7,9]
mi, ma, av = summary(lst)
print(mi, ma, av)

# 5
def sum_all(*args):
    return sum(args)

print(sum_all())
print(sum_all(10))
print(sum_all(1,2,3,4))
```

```python
import os
import pickle

# 1
print("当前工作目录：", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
stu_path = os.path.join(data_dir, "students.txt")

# 2
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
with open(stu_path, "w", encoding="utf-8") as f:
    f.write(content)

valid_list = []
error_log = []

with open(stu_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

# 3、4
from student_utils import parse_student
for idx, line in enumerate(lines[1:], start=2):
    line = line.strip()
    if not line:
        continue
    try:
        rec = parse_student(line)
        valid_list.append(rec)
    except ValueError as e:
        error_log.append((idx, line, str(e)))

print("有效记录：", valid_list)
print("错误记录：")
for no, l, msg in error_log:
    print(f"行{no} | {l} | {msg}")

# 统计
valid_cnt = len(valid_list)
invalid_cnt = len(error_log)
if valid_cnt > 0:
    scores = [d["score"] for d in valid_list]
    avg = sum(scores) / valid_cnt
    pass_num = sum(1 for s in scores if s >= 60)
    pass_rate = pass_num / valid_cnt
else:
    avg = 0
    pass_rate = 0

report_path = os.path.join(data_dir, "report.txt")
# w模式：覆盖清空原有内容
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录：{valid_cnt}\n")
    f.write(f"无效记录：{invalid_cnt}\n")
    f.write(f"平均分：{avg:.2f}\n")
    f.write(f"及格率：{pass_rate:.2%}\n")

# a模式：追加，不会清空旧内容
with open(report_path, "a", encoding="utf-8") as f:
    f.write("=====追加测试(a模式)=====\n")

# 6 pickle
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_list, f)

with open(pkl_path, "rb") as f:
    loaded_data = pickle.load(f)
print("pickle还原校验：", loaded_data == valid_list)

#7 模块导入测试（前面已经import，这里测试）
try:
    print(parse_student("Tom,77"))
except ValueError as e:
    print("有效样例报错：", e)

try:
    print(parse_student("Tom,108"))
except ValueError as e:
    print("无效样例报错：", e)

#8 捕获不存在文件
try:
    with open("no_file.txt","r",encoding="utf-8") as f:
        pass
except FileNotFoundError:
    print("捕获异常：目标文件不存在！")
```

```python
import copy

# 任务1：定义Student类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


# 任务2：使用实验4有效记录创建对象
valid_list = [
    {'name': 'Alice', 'score': 85},
    {'name': 'Bob', 'score': 58},
    {'name': 'Carol', 'score': 92},
    {'name': 'David', 'score': 76},
    {'name': 'Eva', 'score': 60},
    {'name': 'Frank', 'score': 89}
]

print("===== Student 对象信息 =====")
stu_objs = []
for d in valid_list:
    s = Student(d["name"], d["score"])
    stu_objs.append(s)
    status = "及格" if s.is_passed() else "不及格"
    print(f"{s}，状态：{status}")


# 任务3：继承，GraduateStudent子类
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, research_topic={self.research_topic})"

print("\n===== 研究生子类测试 =====")
g = GraduateStudent("Lily", 88, "大数据分析")
print(g)
print("是否及格：", g.is_passed())


# 任务4：直接赋值、浅拷贝copy、深拷贝deepcopy
print("\n===== 拷贝对比实验 =====")
# 原始对象（带嵌套列表）
origin = {"name": "Alice", "scores": [85, 90]}

# 1. 直接赋值
obj1 = origin
obj1["scores"][0] = 0
print("直接赋值后 origin:", origin)
# 恢复原始值，下一组测试重新来
origin = {"name": "Alice", "scores": [85, 90]}

# 2. copy.copy 浅拷贝
obj2 = copy.copy(origin)
obj2["scores"][0] = 0
print("浅拷贝copy.copy后 origin:", origin)
origin = {"name": "Alice", "scores": [85, 90]}

# 3. copy.deepcopy 深拷贝
obj3 = copy.deepcopy(origin)
obj3["scores"][0] = 0
print("深拷贝deepcopy后 origin:", origin)
```
